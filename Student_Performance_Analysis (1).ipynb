# 8: Hypothisis Testing 

H0: There is no significant difference between the mean final grades of male and female students.

H1: There is a significant difference between the mean final grades of male and female students.

Significance level:

α = 0.05

```python
#I perform an independent two-sample t-test to check whether the average final grade differs between male and female students
male_g3 = df[df["sex"] == "M"]["G3"]
female_g3 = df[df["sex"] == "F"]["G3"]

t_stat, p_value = ttest_ind(male_g3, female_g3, equal_var=False)

print("Male mean:", male_g3.mean())
print("Female mean:", female_g3.mean())
print("T-statistic:", t_stat)
print("P-value:", p_value)

alpha = 0.05

if p_value < alpha:
    print("Decision: Reject H0")
else:
    print("Decision: Do not reject H0")
```

# Hypothesis Test Calculation
From the actual dataset:

Male mean G3 ≈ 10.914
Female mean G3 ≈ 9.966
t-statistic ≈ 2.065 in magnitude
p-value ≈ 0.0396
α = 0.05
Because p-value < 0.05, I reject the null hypothesis.

Conclusion
There is a statistically significant difference between the average final grades of male and female students in this dataset. This result shows an association in this sample; it does not prove that gender causes the difference.

# 10: Calculus

# Derivative

```python
# A derivative tells how much an output changes when an input changes.
x = sp.Symbol("x")

y = x**2

derivative = sp.diff(y, x)

print("Function:", y)
print("Derivative:", derivative)
print("Derivative at x=3:", derivative.subs(x, 3))
```

# Partial Derivatives

```python
#When a function has multiple variables, we use partial derivatives.
x, y = sp.symbols("x y")

f = x**2 + y**2

partial_x = sp.diff(f, x)
partial_y = sp.diff(f, y)

print("Function:", f)
print("Partial derivative w.r.t x:", partial_x)
print("Partial derivative w.r.t y:", partial_y)

print("At x=3, y=4:")
print("Partial x:", partial_x.subs({x:3, y:4}))
print("Partial y:", partial_y.subs({x:3, y:4}))
```

# Gradient Descent

```python
#Gradient Descent is an optimization algorithm.
#It minimizes a cost function by moving in the opposite direction of the gradient
x = 10
learning_rate = 0.1

print("Gradient Descent for f(x) = x^2")

for i in range(10):
    gradient = 2 * x
    x = x - learning_rate * gradient
    print(f"Step {i+1}: x = {x}")
```

# Simple Student Performance Gradient Descent
now I apply the same basic idea to student performance.

I use:

G1
G2
Failures
to estimate G3 with a simple linear model.

The model is:

prediction = w1*G1 + w2*G2 + w3*Failures + b

The Mean Squared Error (MSE) is used as the cost function.

The goal is to reduce MSE by updating the weights using gradients.

```python

X = df[["G1", "G2", "failures"]].to_numpy(dtype=float)
y = df["G3"].to_numpy(dtype=float).reshape(-1, 1)

# Add a column of 1s for bias
X_b = np.c_[X, np.ones(len(X))]

weights = np.zeros((4, 1))
learning_rate = 0.0001
epochs = 3000

for i in range(epochs):
    predictions = X_b @ weights
    error = predictions - y
    mse = np.mean(error ** 2)

    gradient = (2 / len(X_b)) * (X_b.T @ error)
    weights = weights - learning_rate * gradient

final_predictions = X_b @ weights
final_mse = np.mean((final_predictions - y) ** 2)

print("Final weights:")
print(weights)
print("\nFinal MSE:", final_mse)
```

 # Cost Function

```python
#Cost function measures how wrong the predictions are.
y_true = np.array([10, 20, 30])
y_pred = np.array([12, 18, 29])

mse = np.mean((y_true - y_pred) ** 2)

print("MSE:", mse)
```
